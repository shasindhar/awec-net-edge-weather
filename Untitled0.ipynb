# 🌦️ AWEC-Net: Weather-Complexity-Aware Adaptive Compression Network
### Full Pipeline — Train → Ablation → Benchmark → Export → Evaluate

**Target**: >97% Validation Accuracy | <50ms CPU Latency | ECE < 0.03

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/shasindhar/awec-net-edge-weather/blob/main/AWEC_Net_Colab.ipynb)

---
**Architecture Highlights**:
- ResNet34 Pretrained Backbone (3-stage multi-exit)
- Lightweight Visual Complexity Estimator G_φ(x) for dynamic routing
- Gumbel-Softmax differentiable routing during training
- Knowledge Distillation from ResNet50 Teacher
- WeightedRandomSampler for Rainy/Snowy/Foggy class balance
- OneCycleLR + Two-Phase LR Optimizer

```
# ── Cell 6: Verify Saved Checkpoint ──────────────────────────────────────────
import os
import torch
from src.config import config
from src.models.awec_net import AWECNet
from src.dataset import get_dataloaders
from src.calibration import compute_ece
import numpy as np

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
ckpt   = os.path.join(config.CHECKPOINT_DIR, 'awec_net_best.pth')

model = AWECNet(
    num_classes=config.NUM_CLASSES,
    pretrained=True,
    backbone_type='resnet34',
    dropout_rate=config.DROPOUT_RATE
).to(device).eval()

if os.path.exists(ckpt):
    model.load_state_dict(torch.load(ckpt, map_location=device))
    print(f'[+] Checkpoint loaded: {ckpt}')
    params = sum(p.numel() for p in model.parameters()) / 1e6
    print(f'[+] Model parameters: {params:.2f}M')

    _, val_loader = get_dataloaders(config.DATA_DIR, batch_size=64, num_workers=4)
    correct, total = 0, 0
    all_probs, all_targets = [], []
    stage_counts = [0, 0, 0]

    with torch.no_grad():
        for images, targets, _ in val_loader:
            images, targets = images.to(device), targets.to(device)
            out = model(images, hard_routing=True)
            probs = torch.softmax(out['logits'], dim=1)
            preds = torch.argmax(probs, dim=1)
            correct += (preds == targets).sum().item()
            total += targets.size(0)
            all_probs.append(probs.cpu().numpy())
            all_targets.append(targets.cpu().numpy())
            routes = torch.argmax(out['routing_weights'], dim=1)
            for r in routes:
                stage_counts[r.item()] += 1

    acc = (correct / total) * 100
    all_probs   = np.concatenate(all_probs)
    all_targets = np.concatenate(all_targets)
    ece = compute_ece(all_probs, all_targets)

    print('='*55)
    print(f'  Final Validation Accuracy : {acc:.2f}%')
    print(f'  ECE Calibration Error     : {ece:.4f}')
    print(f'  Stage Exits (S1/S2/S3)   : {stage_counts[0]}/{stage_counts[1]}/{stage_counts[2]}')
    print('='*55)
else:
    print(f'[!] No checkpoint at {ckpt} — run Cell 5 first')
```

```
# ── Cell 12: Validate ONNX FP32 & INT8 Accuracy ──────────────────────────────
import onnxruntime as ort
import numpy as np
import torch
from src.config import config
from src.dataset import get_dataloaders
from src.calibration import compute_ece

_, val_loader = get_dataloaders(config.DATA_DIR, batch_size=64, num_workers=4)

def onnx_validate(model_path, val_loader, label='FP32'):
    sess = ort.InferenceSession(model_path, providers=['CPUExecutionProvider'])
    in_name = sess.get_inputs()[0].name
    correct, total = 0, 0
    all_probs, all_targets = [], []
    for images, targets, _ in val_loader:
        out = sess.run(None, {in_name: images.numpy()})
        logits = torch.tensor(out[0])
        probs = torch.softmax(logits, dim=1)
        preds = torch.argmax(probs, dim=1)
        correct += (preds == targets).sum().item()
        total += targets.size(0)
        all_probs.append(probs.numpy())
        all_targets.append(targets.numpy())
    acc = (correct / total) * 100 if total > 0 else 0
    all_probs   = np.concatenate(all_probs)
    all_targets = np.concatenate(all_targets)
    ece = compute_ece(all_probs, all_targets)
    size_mb = os.path.getsize(model_path) / 1e6
    print(f'  [{label:10s}] Acc: {acc:.2f}%  ECE: {ece:.4f}  Size: {size_mb:.2f} MB')
    return acc, ece

print('='*55)
print('ONNX Model Validation')
print('='*55)
import os
fp32_path = './checkpoints/onnx/awec_net_fp32.onnx'
int8_path = './checkpoints/onnx/awec_net_int8.onnx'
if os.path.exists(fp32_path):
    onnx_validate(fp32_path, val_loader, 'FP32 ONNX')
if os.path.exists(int8_path):
    onnx_validate(int8_path, val_loader, 'INT8 ONNX')
print('='*55)
```

```
# ── Cell 13: Confusion Matrix & Per-Class Accuracy ───────────────────────────
import torch
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import os
from sklearn.metrics import confusion_matrix, classification_report
from src.config import config
from src.dataset import get_dataloaders
from src.models.awec_net import AWECNet

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = AWECNet(
    num_classes=config.NUM_CLASSES, pretrained=True,
    backbone_type='resnet34', dropout_rate=config.DROPOUT_RATE
).to(device).eval()

ckpt = './checkpoints/awec_net_best.pth'
if os.path.exists(ckpt):
    model.load_state_dict(torch.load(ckpt, map_location=device))

_, val_loader = get_dataloaders(config.DATA_DIR, batch_size=64, num_workers=4)
all_preds, all_targets = [], []

with torch.no_grad():
    for images, targets, _ in val_loader:
        images = images.to(device)
        out = model(images, hard_routing=True)
        preds = torch.argmax(out['logits'], dim=1).cpu()
        all_preds.extend(preds.numpy())
        all_targets.extend(targets.numpy())

all_preds, all_targets = np.array(all_preds), np.array(all_targets)
cm = confusion_matrix(all_targets, all_preds)
cm_norm = cm.astype('float') / cm.sum(axis=1, keepdims=True)

# Print per-class report
print(classification_report(all_targets, all_preds, target_names=config.CLASSES, digits=4))

fig, ax = plt.subplots(figsize=(8, 6))
im = ax.imshow(cm_norm, interpolation='nearest', cmap=plt.cm.Blues)
plt.colorbar(im, ax=ax)
ax.set(xticks=np.arange(len(config.CLASSES)), yticks=np.arange(len(config.CLASSES)),
       xticklabels=config.CLASSES, yticklabels=config.CLASSES,
       ylabel='True Label', xlabel='Predicted Label',
       title='AWEC-Net Confusion Matrix (Normalised)')
plt.setp(ax.get_xticklabels(), rotation=30, ha='right', fontsize=9)

thresh = cm_norm.max() / 2.0
for i in range(len(config.CLASSES)):
    for j in range(len(config.CLASSES)):
        ax.text(j, i, f'{cm_norm[i, j]:.2f}\n({cm[i,j]})',
                ha='center', va='center', fontsize=8,
                color='white' if cm_norm[i, j] > thresh else 'black')

plt.tight_layout()
plt.savefig('./logs/confusion_matrix.png', dpi=150, bbox_inches='tight')
plt.show()
print('[+] Confusion matrix saved → ./logs/confusion_matrix.png')
```

```
# ── Cell 14: Stage Routing Analysis (Complexity Distribution) ─────────────────
import torch
import numpy as np
import matplotlib.pyplot as plt
from collections import defaultdict
from src.config import config
from src.dataset import get_dataloaders
from src.models.awec_net import AWECNet

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = AWECNet(
    num_classes=config.NUM_CLASSES, pretrained=True,
    backbone_type='resnet34', dropout_rate=config.DROPOUT_RATE
).to(device).eval()
ckpt = './checkpoints/awec_net_best.pth'
if os.path.exists(ckpt):
    model.load_state_dict(torch.load(ckpt, map_location=device))

_, val_loader = get_dataloaders(config.DATA_DIR, batch_size=64, num_workers=4)

class_routes   = defaultdict(lambda: [0, 0, 0])
complexity_by_class = defaultdict(list)

with torch.no_grad():
    for images, targets, complexity_cpu in val_loader:
        images = images.to(device)
        out = model(images, hard_routing=True)
        routes = torch.argmax(out['routing_weights'], dim=1).cpu()
        scores = out['complexity_score'].view(-1).cpu().numpy()
        for i, (cls, route, score) in enumerate(zip(targets.numpy(), routes.numpy(), scores)):
            class_routes[cls][route] += 1
            complexity_by_class[cls].append(float(score))

# Plot routing distribution per class
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('AWEC-Net Routing Analysis', fontsize=13, fontweight='bold')

x = np.arange(config.NUM_CLASSES)
s1 = [class_routes[i][0] for i in range(config.NUM_CLASSES)]
s2 = [class_routes[i][1] for i in range(config.NUM_CLASSES)]
s3 = [class_routes[i][2] for i in range(config.NUM_CLASSES)]

axes[0].bar(x, s1, label='Stage 1 (Fast)', color='#2ECC71')
axes[0].bar(x, s2, bottom=s1, label='Stage 2 (Medium)', color='#F39C12')
axes[0].bar(x, [a+b for a,b in zip(s1,s2)], bottom=[a+b for a,b in zip(s1,s2)],
            label='Stage 3 (Deep)', color='#E74C3C')
axes[0].set_xticks(x)
axes[0].set_xticklabels(config.CLASSES)
axes[0].set_title('Stage Routing per Class')
axes[0].set_ylabel('Sample Count')
axes[0].legend()

# Complexity score distribution
for i, cls in enumerate(config.CLASSES):
    scores = complexity_by_class[i]
    if scores:
        axes[1].hist(scores, bins=25, alpha=0.65, label=cls, density=True)
axes[1].set_xlabel('Visual Complexity Score C_ref')
axes[1].set_ylabel('Density')
axes[1].set_title('Complexity Score Distribution per Class')
axes[1].legend(fontsize=8)

plt.tight_layout()
plt.savefig('./logs/routing_analysis.png', dpi=150, bbox_inches='tight')
plt.show()
print('[+] Routing analysis plot saved → ./logs/routing_analysis.png')
```
